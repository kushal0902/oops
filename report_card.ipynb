# Model a Student class with name, roll_number, and marks attributes. Add a method that prints out a formatted report card for the student.

```python
class student :
  def __init__(self, name, roll_no, marks):
    self.name= name
    self.roll_no= roll_no
    self.marks= marks
  def report(self):
    if self.marks>= 90 :
      print(f"{self.name } has secured A+")
    elif self.marks>= 80 :
      print(f"{self.name } has secured A")
    elif self.marks>= 70 :
      print(f"{self.name } has secured B+")
    else:
      print(f"{self.name} is failed")

s1= student("kushal",21,92)
s1.report()
```
