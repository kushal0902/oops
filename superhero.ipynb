# The Superhero: Create a Superhero class. Its constructor (__init__) should take a name and a power. Add a method called use_power() that prints a sentence like: "Spider-Man is using their web-shooting power!" Create an object and call the method.

```python
class superhero:
  def __init__(self, name, power):
    self.name= name
    self.power = power
  def use_power(self):
    print(f"{self.name} is using {self.power}")

s1=superhero("spiderman","webshooter")
s1.use_power()
```
