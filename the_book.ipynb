# The Book (Your Question): Model a Book class with title, author, and price. Add a method called apply_discount that takes a percentage, calculates the new price, and updates the book's price attribute.

```python
class book:
  def __init__(self, title, author, price):
    self.title = title
    self.author = author
    self.price = price
  def discount(self, percentage):
    self.price = self.price * (1-percentage)
    print(f"your discounted price is :{self.price}")

b1=book("harry potter","jk", 5000)
b1.discount(0.1)
```
